```python
import os
from pathlib import Path
import anywidget
import traitlets
import ipywidgets
from oold.ui.anywidget.jsoneditor import JsonEditorWidget

my_schema = {
    "title": "AnyWidget Editor Test",
    "required": ["test"],
    "properties": {"test": {"type": "string"}}
}
editor = JsonEditorWidget(options = {"schema": my_schema})

# log the value of `my_value` any time it changes
def handle_change(change):
    """Prints new value"""
    print("handle_change", change.new)
    editor.options["schema"]["title"] = change.new
    print(editor.options["schema"])
    schema = editor.options["schema"]
    out.clear_output()
    editor = JsonEditorWidget(options = {"schema": schema})
    with out:
        display(editor)
editor.observe(handle_change, "value")

out = ipywidgets.Output()
with out:
    display(editor)
out
```
